 Dataset Acquisition and Description - CA2A - Cillian Murphy x Philip Kelly - ________________ x D00266046

Our dataset is called 'MER Opportunity and Spirit Rovers Pancam Images Labeled Data Set' taken from - https://zenodo.org/records/4302760.
It is based on 3004 iages collected by Pancam instruments which are mounted of the Opportunity and spirit rovers from the NASA Mars Exploration  Rovers Mission. 

There were 25 classes of both scientific and engineering interests. 

```python
# Read in text files with the filenames and labels 
def read_txt_files(path):

    rows = []

    with open(path, encoding='UTF-8') as f:
        for line in f:

            line = line.strip()

            filename = line.split(",")[0]

            features = line.split(",")[1:]

            rows.append({'filename':filename, 'features':features})

        data = pd.DataFrame(rows)

        return data

test_data = read_txt_files('rover_image_classification/mer-labeled-data-set-v1.1.0/test-set-v1.1.0.txt')

train_data = read_txt_files('rover_image_classification/mer-labeled-data-set-v1.1.0/train-set-v1.1.0.txt')

val_data = read_txt_files('rover_image_classification/mer-labeled-data-set-v1.1.0/val-set-v1.1.0.txt')
```
